### Which department has a higher attrition rate?

```python
# Read an image
img = mpimg.imread("Departments.png")

# Display the image
plt.figure(figsize=[18, 25])
plt.imshow(img)

# Remove axis ticks and labels
plt.xticks([])  # Remove x-axis ticks
plt.yticks([])  # Remove y-axis ticks
plt.axis('off') #remove the axis entirely

plt.show()
```

```python
df.Department.nunique()
```

```python
# Prepare Department and Attrition data - Create a Cross Tabulation -
# pd.crosstab() creates a table counting values of Attrition ('Yes' or 'No') for each Department.
# normalize='index' converts the counts into proportions (percent of total per row).
# reset_index() converts the department from index to a regular column for plotting

department_wise_attrition = pd.crosstab(df['Department'], df['Attrition'], normalize='index').reset_index()
department_wise_attrition

#  Convert Proportions to Percentages
department_wise_attrition['No'] = department_wise_attrition['No']*100
department_wise_attrition['Yes'] = department_wise_attrition['Yes']*100
department_wise_attrition
```

```python
pyo.init_notebook_mode(connected=True)

#Rename the columns
department_wise_attrition.columns = ['Department', 'No','Yes']
department_wise_attrition

# Create an interactive bar chart using Plotly
fig = px.bar(department_wise_attrition,
            x='Department',
            y=['No','Yes'],
            barmode='stack',
            labels={'value':'Proportion(%)', 'Department': 'Department'},
            title= 'Department and Attrition'
            )
# Set the content to tooltip
fig.update_traces(hovertemplate="%{X}<br>Proportion:): %{y:.2f}%<extra></extra>", texttemplate = "%{y:.2f}%", textposition='inside')

# Customize the chart layout
fig.update_layout(xaxis_title='Department', yaxis_title='Propotion(%)', title='Department and Attrition')
fig.show()
```

<font color=dodgerblue> 💡 Output: Graph Analysis:</font>
    
Typical Attrition Rates:

- The Research & Development department has the lowest attrition rate, while the Sales department has the highest attrition rate.

### What are the attrition rates by gender?

```python
# Prepare data by gender and attrition status
# Group, unstack, and sort by 'Yes' column in descending order
attrition_by_gender = (df.groupby(['Gender', 'Attrition']).size().unstack().sort_values(by='Yes', ascending=False))

attrition_by_gender
# When to use .unstack():
# After a groupby() with multiple keys (MultiIndex)
# When you want to pivot one of the keys from index to columns
#For making data more readable or easier to visualize (e.g., using plot())
```

```python
# Calculate the total number of employees from each gender
total_employees_by_gender= df['Gender'].value_counts()
total_employees_by_gender
```

```python
# Calculate the number of employees who left from each gender
attrition_counts_by_gender = df[df['Attrition']=='Yes']['Gender'].value_counts()

# Calculate the percentage of employees who left from each gender
attrition_percentage_by_gender = attrition_counts_by_gender/total_employees_by_gender*100
print(attrition_percentage_by_gender)

# Create Horizonantal stacked bar chart
fig, ax = plt.subplots(figsize=[10,6])
attrition_by_gender.plot(kind='barh', stacked=True, color=['#00CED1','#FFA500'], ax=ax)
plt.title('Attrition by Gender')
plt.xlabel('Number of Employees')
plt.ylabel('Gender')

# Annotate the percentages on the bars
for index, value in enumerate(attrition_percentage_by_gender):
    total= attrition_by_gender.sum(axis=1)[index]
    ax.text(total-attrition_by_gender['Yes'][index]/2, index, f'{value:.2f}%', va= 'center', ha='center', color='white', fontsize=12)
plt.show()
```

<font color=dodgerblue> 💡 Output: Graph Analysis:  </font>

- The turnover rate of male employees is higher than that of female employees. This indicates a higher turnover rate among men.

```python
# Create a horizontal stacked bar chart for each gender
fig, ax= plt.subplots(1,2, figsize=(14,8), sharey=True)

# Female employees
attrition_by_gender_marital_status.loc['Female'].plot(kind='barh', stacked=True, color=['#00CED1','pink'], ax=ax[0])
ax[0].set_title('Attrition by Marital Status(Female)')
ax[0].set_xlabel('Number of Employees')
ax[0].set_ylabel('Marital Status')

# Annotate the percentages on the bars
for index, value in enumerate(attrition_percentage_by_gender_marital_status.loc['Female']):
    total = attrition_by_gender_marital_status.loc['Female'].sum(axis=1)[index]
    ax[0].text(total - attrition_by_gender_marital_status.loc['Female', 'Yes'][index] / 2, index, f'{value:.2f}%', va='center', ha='center', color='white', fontsize=12)
    
# Male employees
attrition_by_gender_marital_status.loc['Male'].plot(kind='barh', stacked=True, color=['#00CED1','#FFA500'], ax=ax[1])
ax[1].set_title('Attrition by Marital Status (Male)')
ax[1].set_xlabel('Number of Employees')

# Annotate the percentages on the bars
for index, value in enumerate(attrition_percentage_by_gender_marital_status.loc['Male']):
    total = attrition_by_gender_marital_status.loc['Male'].sum(axis=1)[index]
    ax[1].text(total - attrition_by_gender_marital_status.loc['Male', 'Yes'][index] / 2, index, f'{value:.2f}%', va='center', ha='center', color='white', fontsize=12)
    
plt.tight_layout()
plt.show()    
```

- Single employees have the highest attrition rate in both female and male groups. This may indicate that single employees have a higher tendency to leave the job.
- The attrition rates among married and divorced employees are lower. Particularly, divorced employees have the lowest attrition rate in both gender groups.
- The attrition rate among married employees is slightly higher than that of divorced employees, but still lower compared to single employees.
